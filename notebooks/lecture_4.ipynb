## Example 1: Angular Position of a Pointer

Refer to our solution for Example 1 (Lecture 4). We found out that:

$$f_\Theta(\theta) = \frac{1}{2\pi}$$
$$F_\Theta(\theta) = \frac{\theta}{2\pi}$$

$$P(a < \theta < b) = \int_a^b \frac{1}{2\pi} \, d\theta = \frac{b-a}{2\pi}$$

Run the cell below to visualize the problem.

```python
plot_circle_ex1()
```

We can test our solution by randomly generating a large number of lines from the center, each representing one spin of the pointer landing at a random angle. Then, the probability of the angular position being between a and b is given by:

$$P(a < \theta < b) = \frac{\text{number of lines between positions } a \text{ and } b}{\text{total number of lines}}$$
